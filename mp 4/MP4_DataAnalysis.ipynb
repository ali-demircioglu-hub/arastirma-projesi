```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('data_mp3.csv.csv')
```

```python

print("Mean:\n", data.select_dtypes(include='number').mean())
print("Median:\n", data.select_dtypes(include='number').median())
print("Std:\n", data.select_dtypes(include='number').std())

```

```python
print("Eksik veri:\n", data.isnull().sum())
```

```python


data['Zaman damgası'] = pd.to_datetime(data['Zaman damgası'], errors='coerce')
data['Yıl'] = data['Zaman damgası'].dt.year
data['Ay'] = data['Zaman damgası'].dt.month
data['Gün'] = data['Zaman damgası'].dt.day

```

```python
print("Korelasyon:\n", data.select_dtypes(include='number').corr())
```

```python

numeric_cols = data.select_dtypes(include='number').columns
if len(numeric_cols) > 0:
    sns.histplot(data[numeric_cols[0]], kde=True)
    plt.title('Histogram')
    plt.xlabel(numeric_cols[0])
    plt.ylabel('Frekans')
    plt.show()
else:
    print("Sayısal sütun bulunamadı, histogram çizilemiyor.")
```

```python

numeric_cols = data.select_dtypes(include='number').columns

if len(numeric_cols) >= 2:
    sns.scatterplot(x=numeric_cols[0], y=numeric_cols[1], data=data)
    plt.title('Scatter Plot')
    plt.xlabel(numeric_cols[0])
    plt.ylabel(numeric_cols[1])
    plt.show()
else:
    print("Scatter plot için yeterli sayısal sütun yok.")
```

```python

data['Zaman damgası'] = pd.to_datetime(data['Zaman damgası'], errors='coerce')
data['Yıl'] = data['Zaman damgası'].dt.year
data['Ay'] = data['Zaman damgası'].dt.month
data['Gün'] = data['Zaman damgası'].dt.day
```

```python
print(data.dtypes)
```

```python
print(data.columns)
```

```python

numeric_cols = ['Yıl', 'Ay', 'Gün']
for col in numeric_cols:
    sns.histplot(data[col], kde=False)
    plt.title(f'Histogram: {col}')
    plt.show()
```

```python

categorical_cols = [
    'ne sıklıkla müzik dinlersiniz?',
    'hangi müzik dinleme uygulamasını kullanıyorsunuz?',
    'hangi tür müzikleri ağırlıklı olarak dinlersiniz?'
]

for col in categorical_cols:
    sns.countplot(x=col, data=data)
    plt.title(f'{col} Yanıt Dağılımı')
    plt.xticks(rotation=45)
    plt.show()
```

```python

sns.heatmap(data[['Yıl','Ay','Gün']].corr(), annot=True, cmap='coolwarm')
plt.title('Korelasyon Heatmap')
plt.show()
```
